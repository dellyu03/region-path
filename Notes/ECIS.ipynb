```python
import requests
import pandas as pd
import xml.etree.ElementTree as ET

# =========================================
# API URL
# =========================================

url = "https://eis.work24.go.kr/opi/joApi.do"

# =========================================
# 요청 파라미터
# =========================================
# 종로구 / 남성 / 20~24세 / 2026년 1월 기준

params = {
    "apiSecd": "OPIA",
    "closStdrYm": "202601",
    "rsdAreaCd": "11110",   # 서울 종로구
    "sxdsCd": "M",          # 남성
    "ageCd": "02",          # 20~24세
    "rernSecd": "XML",
    "bgnPage": "1",
    "display": "10"
}

# =========================================
# API 요청
# =========================================

res = requests.get(url, params=params)

print("상태코드:", res.status_code)

# 요청 URL 확인
print("\n요청 URL:")
print(res.url)

# =========================================
# XML 파싱
# =========================================

root = ET.fromstring(res.text)

# =========================================
# 데이터 추출
# =========================================

result_list = []

for item in root.findall(".//rqst"):

    data = {

        # 기준 정보
        "기준년월": item.findtext("dwClosYm"),

        # 지역
        "시도명": item.findtext("ctpvCdNm"),
        "시군구명": item.findtext("rsdAreaCdnm"),

        # 성별 / 연령
        "성별": item.findtext("sxdn"),
        "연령대": item.findtext("ageCdnm"),

        # 업종 / 직종
        "업종명": item.findtext("stdIndLcfn"),
        "직종명": item.findtext("wnetJsfcLrclCdnm"),

        # 수치 데이터
        "신규구인인원수": item.findtext("newJoNmpr"),
        "신규구직건수": item.findtext("newJhntNmpr"),
        "취업건수": item.findtext("empmCt")
    }

    result_list.append(data)

# =========================================
# DataFrame 변환
# =========================================

df = pd.DataFrame(result_list)

# 숫자형 변환
numeric_cols = [
    "신규구인인원수",
    "신규구직건수",
    "취업건수"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# =========================================
# 결과 출력
# =========================================

print("\n===== 종로구 남성 20~24세 고용 데이터 =====")
print(df)

# =========================================
# CSV 저장
# =========================================

df.to_csv(
    "종로구_남성_20_24세_고용데이터.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nCSV 저장 완료")
```
